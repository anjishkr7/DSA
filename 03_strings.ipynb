#Operators on strings

```python
#Aritmetic opr(only + and * works on strings)
print('delhi'+' Mumbai')
print('delhi '*5)
```

```python
#Relational(all works)
'delhi'=='delhi'
```

```python
'mumbai'>'pune'
#lexiographically
```

```python
'Pune'>'pune'
#comparision by UNICODE
```

```python
#LOGICAL
'hello' and 'world'
```

```python
'hello' or 'world'
```

```python
'' and 'world'
```

```python
'' or 'world'
```

```python
not ''
```

```python
not 'world'
```

```python
#loops for string
for i in 'hello':
  print(i)
```

```python
for i in 'delhi':
  print('Pune')
```

```python
#membership opr
'D' in 'delhi'
'D' not in 'delhi'
```
